split test and train data

```python
df=df.sample(frac=0.8,random_state=2)
train_data=df[:17000]
train_data=train_data.reset_index(drop=True)
test_data=df[17000:]
test_data=test_data.reset_index(drop=True)
```

```python
y_train=train_data['median_house_value'].to_numpy()
x_train=train_data['median_house_value'].to_numpy()
```

Get error of baseline model that uses the avarage median house value in train data

```python
avarage_median_house_value=train_data['median_house_value'].mean()
```

```python
baseline_model_test_prediction =[avarage_median_house_value]* len(y_train)
```

```python
mean_squared_error(baseline_model_test_prediction,x_train)
```
