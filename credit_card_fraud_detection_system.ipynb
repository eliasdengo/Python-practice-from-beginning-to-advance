take sample data

```python
legit_sample=legit.sample(n=204)
```

```python
fraud_sample=fraud.sample(n=204)
```

merege sample data

```python
mereged_data=pd.concat([legit_sample,fraud_sample],axis=0)
```

```python
mereged_data.shape
```

split feature and target data

```python
x=mereged_data.drop(columns='Class',axis=1)
y=mereged_data['Class']
```

split data for test and train model

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
x.shape,x_train.shape,x_test.shape
```

train machine learning model

```python
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

model evaluation

```python
x_train_prediction=model.predict(x_train)
```

```python
x_train_prediction_accuracy=accuracy_score(x_train_prediction,y_train)
```

```python
print('accuracy on training data',x_train_prediction_accuracy)
```
