```python
#converting to numerical value
house_price_dataset.replace({"mainroad":{"no":0,"yes":1},"guestroom":{"yes":1,"no":0},"basement":{"yes":1,"no":0},"hotwaterheating":{"yes":1,"no":0},"airconditioning":{"no":0,"yes":1},"prefarea":{"no":0,"yes":1},"furnishingstatus":{"unfurnished":0,"semi-furnished":1,"furnished":2}},inplace=True)
```

```python
correlation=house_price_dataset.corr()
```

constructing heatmap to understand correlation

```python
plt.figure(figsize=(10,10))
sns.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Greens')
```

splitting the data and target

```python
y=house_price_dataset['price']
x=house_price_dataset.drop(['price'],axis=1)
```

```python
print(x)
print(y)
```

spiliting data into test and train data

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

Model Training

XGBoost Regressor

```python
#loading models
model=XGBRegressor()
```

```python
#training mode x_train and y_train
model.fit(x_train,y_train)
```

Model Evaluation


predicition in training data


```python
#accuracy for prediction on training data
training_data_prediction=model.predict(x_train)
```

```python
print(training_data_prediction)
```

```python
# R square error
score_1=metrics.r2_score(y_train,training_data_prediction)
```

```python
# mean absolute error
score_2=metrics.mean_absolute_error(y_train,training_data_prediction)
```

```python
print('R square error:',score_1)
print('Mean absolute error:',score_2)
```

Test data prediction

```python
#accuracy for prediction on training data
test_data_prediction=model.predict(x_test)
```

```python
print(test_data_prediction)
```

```python
# R square error
score_1=metrics.r2_score(y_test,test_data_prediction)
```

```python
# mean absolute error
score_2=metrics.mean_absolute_error(y_test,test_data_prediction)
```

```python
print('R square error:',score_1)
print('Mean absolute error:',score_2)
```
